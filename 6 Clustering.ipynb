```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\sgban\Desktop\BackML\6\Mall_Customers.csv")
```

```python
df.head()
```

```python
X = df.iloc[:, 3:]
X
```

```python
plt.title('Unclustered Data')
plt.xlabel('Annual Income')
plt.ylabel('Spending Income')
plt.scatter(X['Annual Income (k$)'], X['Spending Score (1-100)'])
```

```python
from sklearn.cluster import KMeans, AgglomerativeClustering
km = KMeans(n_clusters=3)
km.fit(X)
```

```python
pred = km.predict(X)
print(pred)
```

```python
elbow = []
for k in range(1,16):
  km = KMeans(n_clusters = k)
  km.fit_predict(X)
  elbow.append(km.inertia_)
```

```python
elbow
```

```python
import matplotlib.pyplot as plt
plt.xticks(range(1,16))
plt.plot(range(1,16), elbow)
plt.xlabel('K')
plt.ylabel('Inertia')
plt.show()
```

```python
km1 = KMeans(n_clusters = 5)
km1.fit_predict(X)
```

```python
plt.figure(figsize = (16,9))
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.scatter(X.iloc[:,0], X.iloc[:,1], c = km1.labels_)
plt.scatter(km1.cluster_centers_[:, 0], km1.cluster_centers_[:,1], color = 'r')

plt.show()
```

```python
from sklearn.metrics import silhouette_score
agg = []
for k in range(2,16):
  ag = AgglomerativeClustering(n_clusters = k)
  pred = ag.fit_predict(X)
  agg.append(silhouette_score(X, pred))
```

```python
plt.plot(range(2,16), agg)
plt.xticks(range(2,16))
plt.show()
```

```python

ag1 = AgglomerativeClustering(n_clusters = 5)
pred1 = ag1.fit_predict(X)
import numpy as np
colors = ['purple', 'blue', 'green', 'yellow', 'black']

# Create a scatter plot using a loop
for i in range(len(X)):
    plt.scatter(X.iloc[i, 0], X.iloc[i, 1], color=colors[pred1[i]])

plt.scatter(km1.cluster_centers_[:,0], km1.cluster_centers_[:,1], color = 'red')
plt.show()
```

```python
plt.figure(figsize=(16,9))
plt.subplot(1,2,1)
plt.title('Unclustered Data')
plt.xlabel('Annual Income')
plt.ylabel('Spending Income')
plt.scatter(X['Annual Income (k$)'], X['Spending Score (1-100)'])

plt.subplot(1,2,2)
plt.title('Clustered Data')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.scatter(X.iloc[:,0], X.iloc[:,1], c = km1.labels_)
plt.scatter(km1.cluster_centers_[:, 0], km1.cluster_centers_[:,1], color = 'r')

plt.show()
```
