```python
import pandas as pd
import csv
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
dataset = []
with open(r"C:\Users\sgban\Desktop\BackML\7\Market_Basket_Optimisation.csv") as file:
    reader = csv.reader(file, delimiter=',')
    for row in reader:
        dataset+=[row]
```

```python
dataset[0:10]
```

```python
len(dataset)
```

```python
te = TransactionEncoder()
x = te.fit_transform(dataset)
```

```python
x
```

```python
te.columns_
```

```python
df = pd.DataFrame(x, columns=te.columns_)
```

```python
df
```

```python
# Finding frequent itemsets
freq_itemset = apriori(df, min_support=0.01, use_colnames=True)
freq_itemset
```

```python
#Find the rules
rules = association_rules(freq_itemset, metric='confidence', min_threshold=0.25)
```

```python
rules = rules[['antecedents','consequents','support','confidence']]
rules
```
